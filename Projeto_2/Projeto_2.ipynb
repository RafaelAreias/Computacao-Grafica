 Passo 1: Importar as bibliotecas 

```python
import moviepy.editor as mpe
import numpy as np
import matplotlib.pyplot as plt
```

Passo 2: Abrir o vídeo original


```python
video = mpe.VideoFileClip("Trabalho.mp4")
```

Passo 3: Inverter a imagem no eixo a cada 20 segundos

```python
def inverter_eixo(get_frame, t):
    frame = get_frame(t)
    if (int(t) // 20) % 2 == 1:    # Divide o tempo em intervalos de 20 segundos
        return frame[:, ::-1, :]   # Inverte horizontalmente
    return frame

video_invertido = video.fl(inverter_eixo)
video_invertido.ipython_display(width=480, maxduration=160)
```

Passo 4: Diminuir gradativamente o som a cada 30 segundos

```python
def volume(t):                                                      # Função recebe o tempo t
    t = np.asarray(t)
    intervalos = np.array([0, 30, 60, 90, 120, 150, np.inf])        
    intensidades = np.array([1.0, 0.8, 0.6, 0.4, 0.2, 0.0])
    return intensidades[np.digitize(t, intervalos) - 1]             # Retorna a intensidade correspondente ao indice do intervalo de t

novo_audio = video_invertido.audio.fl(lambda gf, t: gf(t) * volume(t)[:, np.newaxis])   # Aplica a função volume ao áudio
audio_diminuido = video_invertido.set_audio(novo_audio)                                 # Substitui o áudio original pelo novo áudio com volume modificado
audio_diminuido.ipython_display(width=480, maxduration=160)
```

Passo 5: Corte do segundo 60 para o segundo 80

```python
corte_inicio = audio_diminuido.subclip(0, 60) # Corta os primeiros 60 segundos
corte_fim = audio_diminuido.subclip(80, audio_diminuido.duration)  # Corta do segundo 80 até o fim do vídeo

cortado = mpe.concatenate_videoclips([corte_inicio, corte_fim])  # Concatena os cortes
cortado.ipython_display(width=480, maxduration=160)
```

Passo 6: Inserir os 20 segundos cortado no fim do vídeo

```python
corte = audio_diminuido.subclip(60, 80)                       # Corta do segundo 60 ao 80
video_final = mpe.concatenate_videoclips([cortado, corte])    # Concatena o resulatdo anterior com o corte do meio
video_final.ipython_display(width=480, maxduration=160) 
video_final.write_videofile("Video_final.mp4")                # Salva o vídeo final
```
